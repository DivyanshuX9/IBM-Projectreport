## Question 1 — Check if a Sentence is Positive or Negative

```python
!pip install transformers -q
```

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

sentences = [
    "I love this product, it is amazing!",
    "This is the worst experience I have ever had.",
    "The food was really good and tasty.",
    "I am very disappointed with the service."
]

results = classifier(sentences)

for i, (sentence, result) in enumerate(zip(sentences, results), start=1):
    print(f"Sentence {i}: {sentence}")
    print(f"  Result : {result['label']}")
    print(f"  Score  : {result['score']:.4f}")
    print()
```

## Question 2 — Complete a Sentence using GPT-2

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

prompts = [
    "Artificial intelligence is",
    "The best way to learn programming is",
    "In the future, robots will"
]

for prompt in prompts:
    print(f"Prompt: {prompt}")
    completions = generator(
        prompt,
        max_new_tokens=40,
        num_return_sequences=2,
        do_sample=True,
        temperature=0.8
    )
    for j, completion in enumerate(completions, start=1):
        print(f"  Completion {j}: {completion['generated_text']}")
    print()
```

## Question 3 — Predict the Missing Word in a Sentence

```python
from transformers import pipeline

fill_mask = pipeline("fill-mask", model="bert-base-uncased")

sentences = [
    "The student is studying for the [MASK] exam.",
    "She cooked a delicious [MASK] for dinner.",
    "He went to the [MASK] to buy vegetables.",
    "The doctor gave the patient a [MASK]."
]

for sentence in sentences:
    print(f"Sentence: {sentence}")
    predictions = fill_mask(sentence, top_k=3)
    for k, pred in enumerate(predictions, start=1):
        word = pred['token_str']
        score = pred['score'] * 100
        print(f"  Prediction {k}: {word} — {score:.2f}%")
    print()
```

## Question 4 — Count How Many Tokens are in a Sentence

```python
!pip install tiktoken -q
```

```python
import tiktoken

enc = tiktoken.encoding_for_model("gpt-4")

sentences = [
    "Hello, how are you?",
    "Machine learning is a part of artificial intelligence.",
    "India is a beautiful country with many languages and cultures."
]

for sentence in sentences:
    tokens = enc.encode(sentence)
    decoded = enc.decode(tokens)
    print(f"Sentence : {sentence}")
    print(f"Tokens   : {tokens}")
    print(f"Count    : {len(tokens)} tokens")
    print(f"Decoded  : {decoded}")
    print()
```

## Question 5 — Search for Top AI Models on Hugging Face

```python
!pip install huggingface_hub -q
```

```python
from huggingface_hub import HfApi

api = HfApi()

tasks = [
    "text-generation",
    "sentiment-analysis"
]

for task in tasks:
    print(f"Task: {task}")
    models = list(api.list_models(pipeline_tag=task, sort="downloads", limit=3))
    for rank, m in enumerate(models, start=1):
        downloads = getattr(m, "downloads", 0) or 0
        print(f"  Rank {rank} : {m.modelId} — {downloads:,} downloads")
    print()
```
